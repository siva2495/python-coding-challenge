```python
def merge_the_tools(string, k):
    factor = len(string)//k
    for i in range (factor):
        start = i*k
        end = start+k
        u = string[start: end]
        
        result = ""
        for ch in u:
            if ch not in result:
                result +=ch
        print(result)
        

if __name__ == '__main__':
    string, k = input(), int(input())
    merge_the_tools(string, k)
```
