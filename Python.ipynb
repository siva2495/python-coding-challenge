# Compress String
input s = "aaabccdddd"
output = "a3b1c2d4"

```python
def compress_string(s):
    result =""
    count = 1
    
    for i in range(1, len(s)):
        if s[i] == s[i - 1]:
            count += 1
        else:
            result += s[i - 1] + str(count)
            count = 1
    result += s[-1] + str(count)
    return result

print(compress_string("aaabccdddd"))  # Output: "a3b3c2"
    
```

# Merge Intervals
input: [[1,3], [2,6], [8,10], [15,16]]
output: [[1,6], [8,10], [15, 16]]

```python
def merge_intervals(intervals):
    if not intervals:
        return []
    
    intervals.sort(key=lambda x: x[0])
    merged = []
    
    for interval in intervals:
        if not merged or  merged[-1][1] < interval[0]:
            merged.append(interval)
        else:
            merged[-1][1] = max(merged[-1][1], interval[1])
    
    return merged
```

# Top k words:
k=2
input = "the cat and the dog and the bird"
ouptut = ((the,3), (and,2))

```python
def top_k_frequent(text, k):
    words = text.lower().split()
    count = {}
    
    for word in words:
        if word in count:
            count[word] += 1
        else:
            count[word] = 1
            
    sorted_words = sorted(count.items(), key=lambda x: x[1], reverse=True)
    return sorted_words[:k]

text = "This is a test. This test is only a test."
k = 2
print(top_k_frequent(text, k))  # Output: [('test.', 3), ('this', 2)]
```
