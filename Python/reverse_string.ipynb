# Reverse String Without using Slicing [::-1] and reversed

```python
def reverse_string(s):
    result = ""
    for char in s:
        result = char + result
    return result

print(reverse_string("google"))
```

# Reverse string using slicing

```python
def reverse_sring(s):
    result = ""
    for char in s:
        result = s[::-1]
    return result

print(reverse_string("google"))
```

# Resverse String using reversed()

```python
s = "google"
reversed_chars = list(reversed(s))
print("".join(reversed_chars))
```
