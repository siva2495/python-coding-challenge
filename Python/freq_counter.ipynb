# Count Character Frequencies

```python
def char_freqncy(s):
    freq = {}
    for char in s:
        if char in freq:
            freq[char] += 1
        else:
            freq[char] = 1
    return freq

print(char_freqncy("google"))
```

* Using dict.get(char, 0)

```python
def char_frequency(s):
    freq = {}
    for char in s:
        freq[char] = freq.get(char, 0) + 1
    return freq

print(char_frequency("google"))
```
