* IS Palindrome ("madam" reads same forward and backward)

* using slice

```python
def is_palindrome(s):
    s = s.lower()
    return s == s[::-1]
print(is_palindrome("madam"))
```

```python
print(is_palindrome("hello"))
```

* Using While loop

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1
    while left < right:
        if s[left].lower() != s[right].lower():
            return False
        left += 1
        right -= 1
    return True

print(is_palindrome("Madam"))
```

```python
print(is_palindrome("Hello"))
```
