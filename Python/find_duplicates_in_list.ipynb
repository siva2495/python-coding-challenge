* Find Duplicates

```python
def find_duplicates(nums):
    seen = set()
    duplicates = set()
    for num in nums:
        if num in seen:
            duplicates.add(num)
        else:
            seen.add(num)
    return list(duplicates)

print(find_duplicates([1, 2, 3, 2, 4, 5, 1]))  # Output: [1, 2]
```

* Remove duplicates While preserving order

```python
def remove_duplicates(nums):
    return list(set(nums))

print(remove_duplicates([1, 2, 3, 2, 4, 5, 1]))  # Output: [1, 2, 3, 4, 5]
```

```python
def remove_duplicates(nums):
    seen = set()
    result = []
    for num in nums:
        if num not in seen:
            seen.add(num)
            result.append(num)
    return result

print(remove_duplicates([1, 2, 3, 2, 4, 5, 1]))  # Output: [1, 2, 3, 4, 5]
```
