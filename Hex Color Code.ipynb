```python
import re
lines = [
    "11",
    "#BED",
    "{",
    "    color: #FfFdF8;",
    "    background-color:#aef;",
    "    border: 2px dashed #fff;",
    "}",
]
n = int(lines[0])

for _ in range(n):
    line = input()

    matches = re.findall(r'(?<=[: ,])#(?:[0-9a-fA-F]{3}|[0-9a-fA-F]{6})\b', line)

    for match in matches:
        print(match)
```
